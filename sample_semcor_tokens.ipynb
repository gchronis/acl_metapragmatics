```python
from nltk.tree import Tree

import pandas as pd

from nltk.corpus.reader.wordnet import Lemma
from nltk.corpus import wordnet as wn
from nltk.stem import WordNetLemmatizer
import numpy as np


from scipy.stats import spearmanr, pearsonr

```

```python
tokens_df = pd.read_csv('./collected_tokens/semcor/all_tokens.csv')
tokens_df
```

```python

# Group by lemma and sense, then count distinct token_ids
grouped = tokens_df.groupby(['lemma', 'sense'])['sentence_id'].nunique().reset_index()

# Keep only lemma groups with at least 5 distinct token_ids per sense
valid_senses = grouped[grouped['sentence_id'] >= 20]

# Count how many senses each lemma has after filtering
lemma_counts = valid_senses.groupby('lemma')['sense'].nunique()

# Keep only lemmas that have at least two distinct senses
valid_lemmas = lemma_counts[lemma_counts >= 2].index


# Filter the original DataFrame
filtered_df = tokens_df[tokens_df['lemma'].isin(valid_lemmas)]

good_lemmas = filtered_df['lemma'].unique()
len(good_lemmas)
```

```python

# Filter the DataFrame to keep only the selected lemmas
polysemous_sample = filtered_df[filtered_df['lemma'].isin(selected_lemmas)]
len(polysemous_sample)
```
