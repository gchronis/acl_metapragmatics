```python
top_df = top_bottom_sents[top_bottom_sents['top_bottom'] == 'top']
```

```python
from bertopic import BERTopic
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

docs = top_df['sentence']          
topics = top_df['feature']  

# Convert string labels to integers 
unique_labels = list(set(topics))       # List of predefined topic labels for each doc (ints or strings)
label_to_int = {label: idx for idx, label in enumerate(unique_labels)}
int_to_label = {idx: label for idx, label in enumerate(unique_labels)}
int_topics = [label_to_int[label] for label in topics]


# Example:
# docs = ["text1", "text2", "text3", ...]
# topics = ["topic_A", "topic_B", "topic_A", ...]

# Create a dummy BERTopic model to access c-TF-IDF logic
model = BERTopic()

# Fit only to compute the vectorizer and embeddings (optional but helpful)
#model.fit_transform(docs)

# Replace model's topics with your own
model.custom_labels_ = topics
model.topics_ = topics


model.update_topics(docs, topics=int_topics)
```

```python
topic_info = model.get_topic_info()
#model.get_topic("topic_A")   # or model.get_topic(0) if labels are ints

```

```python
topic_info['original_label'] = topic_info['Topic'].map(int_to_label)
topic_df = topic_info[['original_label', 'Name']]
topic_df
```
