```python
# list of features with less than 10 occurrences
low_freq_features_10 = df['translated'].value_counts()[df['translated'].value_counts() < 10].index.tolist()
print("Number of features occurring less than 10 times: ", len(low_freq_features_10))

# list of features with less than 5 occurrences
low_freq_features_5 = df['translated'].value_counts()[df['translated'].value_counts() < 5].index.tolist()
print("Number of features occurring less than 5 times: ", len(low_freq_features_5))
```

```python
# save low frequency features to a file
with open('/home/gsc685/data/buchanan_ratings/low_frequency_features_<10.txt', 'w') as f:
    for feature in low_freq_features_10:
        f.write("%s\n" % feature)

# save low frequency features to a file
with open('/home/gsc685/data/buchanan_ratings/low_frequency_features_<5.txt', 'w') as f:
    for feature in low_freq_features_5:
        f.write("%s\n" % feature)
```
